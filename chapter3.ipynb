```python
emp = spark.createDataFrame(data=emp_data, schema=emp_schema)
```

```python
emp.show()
```

```python
emp.printSchema()
```

```python
from pyspark.sql.functions import col,cast

emp_casted = emp.select("employee_id", "name", "age", "gender", col("salary").cast("double").alias("salary"), col("hire_date").cast("date").alias("hire_date"))
emp_casted.show()
```

```python
emp_casted.printSchema()
```

```python
# to add or overwrite a column, we can use withColumn() method.
# It takes two arguments, the name of the column and the value to be assigned to that column. 

emp_taxed = emp_casted.withColumn("tax", col("salary") * 0.2)
emp_taxed.show()
```

```python
# how to add a constant column to a DataFrame using lit() function. 

from pyspark.sql.functions import lit

emp_constant = emp_taxed.withColumn("constant_value", lit(1)).withColumn("constant_string", lit("constant"))
emp_constant.show()
```

```python
# add multiple columns at once
coloum = {
    "tax": col("salary") * 0.2,
    "constant_value": lit(1),
    "constant_string": lit("constant")
}

emp_final = emp.withColumns(coloum)
emp_final.show()
```

```python
# for renaming a column
emp_renamed = emp_constant.withColumnRenamed("employee_id", "emp_id")
emp_renamed.show()
```

```python
# to drop a multiple columns 
emp_dropped = emp_constant.drop("constant_value", "constant_string")
emp_dropped.show()
```

```python
emp_filtered = emp_dropped.where("tax > 10000")
emp_filtered.show()
```

```python
# how to limit the result
emp_limited = emp_filtered.limit(5)
emp_limited.show()
```

```python
# can also limit using show()
emp_filtered.show(2)
```
